## Task 3: Broadcasting(1 Mark)

Perform broadcasting operations on matrix A created in Task-2.

Add scalar 10 to matrix A

Add vector [1,2,3,4,5] to each row

Normalize matrix values between 0–1

```python
# Perform broadcasting operations here
a =np.random.randint(1,2,(5,5))
print(a)

# Add scalar 10 to matrix
a_add = a +10
print("\n after adding 10")
print(a_add)

# Add vector [1,2,3,4,5] to each row
vector = [1,2,3,4,5]
a_vector = a+vector
print("\n adding the matrix with vector [1,2,3,4,5]")
print(a_vector)

# Normalize matrix between 0 and 1
a_max = a.max()
a_min =a.min()
a_normalize = (a - a_min)/(a_max -a_min)

print("\n normalize matrix")
print("\n")
print(a_normalize)
```

## Task 4: Vectorized Computation(1 Marks)

Perform element-wise multiplication
Compute dot product
Compute mean, standard deviation

```python
X = np.array([1,2,3,4])
Y = np.array([5,6,7,8])

# Element-wise multiplication

A_mutilplication = X * Y
print("\n elemnt wise mutiplication")
print(A_mutilplication)

# Dot product

dot_product = np.dot(X, Y)
print("\n Dot Product:")
print(dot_product)

# Mean and Standard Deviation

mean_X= np.mean(X)
std_X= np.std(X)
print("\n Mean of X:", mean_X)
print("\n")
print("Standard Deviation of X:", std_X)
```

Task 5: Mini Case Problem (1 Mark)

Dataset: Student marks (5 students, 3 subjects)

Compute total and average per student

Find topper

Find subject-wise highest score

```python
marks = np.array([[80,75,90],[60,88,70],[95,92,85],[70,65,60],[88,78,84]])
print("marks")
print(marks)
# Compute totals, averages, topper
#total and average
total = np.sum(marks, axis=1)
average = np.mean(marks,axis=1)
print("\nTotal Marks per Student:")
print(total)

print("\nAverage Marks per Student:")
print(average)

# topper
topper_index = np.argmax(total)
print("\nTopper is Student", topper_index + 1)
print("Topper Total Marks:", total[topper_index])

```
